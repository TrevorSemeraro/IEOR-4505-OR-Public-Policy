```python
import geopandas as gpd
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
tracts = gpd.read_file("./1980/NY_block_1980/NY_block_1980.shp")
zip_codes = gpd.read_file("zipcode/geo_export_b459f634-6bcb-46d7-ade3-b0ef95b76ebc.shp")
```

```python
manhattan = tracts[tracts["COUNTY"] == "New York"]
```

```python
# Ensure CRS is WGS84 (EPSG:4326) for web-based plotting
manhattan = manhattan.to_crs(epsg=4326)

# Plot using Plotly Express
fig = px.choropleth_mapbox(
    manhattan,
    geojson=manhattan.geometry,
    locations=manhattan.index,
    color_discrete_sequence=["white"],
    hover_data={"TRACTA": True},
    center={"lat": manhattan.geometry.centroid.y.mean(), "lon": manhattan.geometry.centroid.x.mean()},
    zoom=11,
    mapbox_style="carto-positron",
)

fig.update_traces(marker_line_width=1, marker_line_color="black")
fig.update_layout(
    margin={"r": 0, "t": 30, "l": 0, "b": 0},
    title_text="1980 Census Tracts (NY)",
    title_x=0.5
)

fig.show()
```

```python
df_1980 = pd.read_csv("./1980/1980.csv")
df_1980.fillna(0, inplace=True)
df_1980
```

```python
def get_perc_populations(df):
    total_white = df["white"].sum()
    total_black = df["black"].sum()
    total_asian = df["asian"].sum()

    total = total_white + total_black + total_asian
    perc_white = total_white / total * 100
    perc_black = total_black / total * 100
    perc_asian = total_asian / total * 100

    print(f"White: {perc_white:.2f}%")
    print(f"Black: {perc_black:.2f}%")
    print(f"Asian: {perc_asian:.2f}%")
    
    return perc_white, perc_black, perc_asian
```

```python
# many pops
tracts_10019 = [133, 135, 136, 139]
tracts_10017 = [90, 91, 93, 97]
# some pops (Lower East Side)
tracts_10002 = [8, 16, 25]
# no pops (right above central park)
tracts_10026 = [186, 190, 216, 218]

df_1980_10019 = df_1980[df_1980["tract"].isin(tracts_10019)]
df_1980_10017 = df_1980[df_1980["tract"].isin(tracts_10017)]
df_1980_10002 = df_1980[df_1980["tract"].isin(tracts_10002)]
df_1980_10026 = df_1980[df_1980["tract"].isin(tracts_10026)]
```

```python
get_perc_populations(df_1980_10019)
get_perc_populations(df_1980_10017)
```
