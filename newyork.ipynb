```python
import pandas as pd

input_file = "fy2025_safmrs.csv"
output_file = "filtered_new_york.csv"

# Define chunk size
chunk_size = 5000

# Initialize an empty DataFrame to store results
filtered_data = []

# Read the CSV file in chunks
for chunk in pd.read_csv(input_file, chunksize=chunk_size, dtype=str):
    ny_rows = chunk[chunk["HUD Fair Market Rent Area Name"].str.contains(" NY ", case=False, na=False)]
    
    # Append filtered rows to the list
    filtered_data.append(ny_rows)

# Concatenate all filtered data and save to a new CSV file
if filtered_data:
    pd.concat(filtered_data).to_csv(output_file, index=False)
    print(f"Filtered data saved to {output_file}")
else:
    print("No matching records found.")
```
