```python
import os
import pandas as pd
import numpy as np
```

```python
racial_2000 = pd.read_csv(os.path.join('racial_data', '2000data.csv'))
racial_2010 = pd.read_csv(os.path.join('racial_data', '2010data.csv'))
racial_2020 = pd.read_csv(os.path.join('racial_data', '2020data.csv'))

racial_2000["White alone"] = racial_2000["White alone"].str.replace(",", "").astype(float)
racial_2000["Black or African American alone"] = racial_2000["Black or African American alone"].str.replace(",", "").astype(float)
racial_2000["Asian alone"] = racial_2000["Asian alone"].str.replace(",", "").astype(float)
racial_2000["Total:"] = racial_2000["Total:"].str.replace(",", "").astype(float)
```

```python
racial_2000["ZIP"] = racial_2000["NAME"].str.split(" ").str[-1]
racial_2000
```

```python
total_sum_2000 = racial_2000["Total:"].sum()
white_sum_2000 = racial_2000["White alone"].sum()
black_sum_2000 = racial_2000["Black or African American alone"].sum()
asian_sum_2000 = racial_2000["Asian alone"].sum()

total_sum_2020 = racial_2020["P8_001N"]
white_sum_2020 = racial_2020["P8_003N"]
black_sum_2020 = racial_2020["P8_004N"]
asian_sum_2020 = racial_2020["P8_006N"]
```

```python
print(white_sum_2000 / total_sum_2000)
```

```python
racial_2000["perc_white"] = racial_2000["White alone"] / racial_2000["Total:"]
racial_2000["perc_black"] = racial_2000["Black or African American alone"] / racial_2000["Total:"]

racial_2000["perc_white"].mean()
```

```python
# filter zip codes below 110th streets
allowed_zip_codes = [
    "10025",
    "10024",
    "10023",
    "10019",
    "10036",
    "10018",
    "10001",
    "10011",
    "10014",
    "10012",
    "10013",
    "10007",
    "10038",
    "10004",
    "10005",
    "10006",
    "10048",
]

below_racial_2000_filtered = racial_2000[racial_2000["ZIP"].isin(allowed_zip_codes)]
above_racial_2000_filtered = racial_2000[~racial_2000["ZIP"].isin(allowed_zip_codes)]
```
