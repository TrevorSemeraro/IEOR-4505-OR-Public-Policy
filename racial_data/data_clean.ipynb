```python
import pandas as pd

def transpose_csv(input_file, output_file):
    # Load transposed CSV into pandas DataFrame
    df = pd.read_csv(input_file)
    
    # Trim column titles
    df.columns = df.columns.str.strip()
    
    # Save final CSV file
    df.to_csv(output_file, index=False)
    
    print(f"Final cleaned CSV saved as {output_file}")

# Example usage
if __name__ == "__main__":
    input_csv = "./racial_data/2000dataf.csv"  # Replace with your input file
    output_csv = "./racial_data/2000dataf2.csv"  # Replace with your output file
    transpose_csv(input_csv, output_csv)
    print(f"Transposed CSV saved as {output_csv}")
```
